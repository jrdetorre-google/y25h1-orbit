# Step 1: Create a Spark session

```python
from google.cloud.spark_connect import GoogleSparkSession
from google.cloud.dataproc_v1 import Session
```

Create your interactive Spark Session with one single line

```python
spark = GoogleSparkSession.builder.getOrCreate()
```

Use a customized Spark Session

```python
# If you would like to customize the Spark session, please refer to the
# documentation at https://cloud.google.com/bigquery/docs/use-spark
#
# For example, if you need to use a different network, use the code below

session = Session()
session.environment_config.execution_config.network_uri = "vpc-main"
session.environment_config.execution_config.subnetwork_uri = "dataproc-serverless-subnet"
spark = (
    GoogleSparkSession.builder
      .appName("test")
      .googleSessionConfig(session)
      .getOrCreate()
)
```

# Step 2: Load data

Let's load the products and orders data.

```python
# Load data from BigQuery
products = spark.read.format('bigquery') \
  .option('table', 'bigquery-public-data.thelook_ecommerce.products') \
  .load()
products.createOrReplaceTempView('products')
```

```python
order_items = spark.read.format('bigquery') \
  .option('table', 'bigquery-public-data.thelook_ecommerce.order_items') \
  .load()
order_items.createOrReplaceTempView('order_items')
```
